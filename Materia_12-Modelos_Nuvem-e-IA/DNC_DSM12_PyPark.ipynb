```python
from pyspark import SparkContext
```

```python
sc = SparkContext("local", "RDD Exemplo")
```

```python
# Leia o arquivo com a codificação UTF-8
with open("C:/Users/bianc/OneDrive/Documents/1-Estudos/2-DNC/1-Data Science/1-Material/Materia12-Modelos_de_Nuvem-e-IA/Aula 5/rdd_dataset.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

# Crie o RDD e limpe as strings em uma única operação
rdd_texto = sc.parallelize(lines).map(lambda linha: linha.strip())

# Verifique os resultados
palavras = rdd_texto.collect()
print(palavras)
```

```python
rdd_texto.collect()
```

```python
rdd_filtrado = rdd_texto.filter(lambda linha: "DNC" in linha)
```

```python
rdd_filtrado.collect()
```

```python
rdd_filtrado.count()
```

```python
df = spark.read.csv(
      path="C:/Users/bianc/OneDrive/Documents/1-Estudos/2-DNC/1-Data Science/1-Material/Materia12-Modelos_de_Nuvem-e-IA/Aula 6/base_de_dados.csv",
      header=True,
      sep=";"
)
```

```python
from pyspark.sql.functions import col, to_timestamp
```

```python
df_cast = df.withColumn(
    'id', col('id').cast('int'),
).withColumn(
    'valor', col('valor').cast(DoubleType()),
).withColumn(
    'data_transacao', to_timestamp('data_transacao', "dd/MM/yyyy HH:mm"),
)
```

```python
df_cast.printSchema()
```

```python
df_cast.show()
```

### Adicionando mais colunas

Exemplo para transformar coluna de pix para dolar

```python
from pyspark.sql.functions import col, round, lit

dolar_hoje = 5.15
df_dolar = df.withColumn(
    'dolar',lit(dolar_hoje)
).withColumn(
    'valor_dolar', round(col('valor') * col('dolar'), 2)
)
```

```python
df_dolar.select('id', 'valor', 'dolar', 'valor_dolar').show()
```

### Dropar Colunas

```python
df_dolar = df_dolar.drop('dolar')
```

```python
df_dolar.printSchema()
```

### Renomear Colunas

```python
df_dolar = df_dolar.withColumnRenamed('valor_dolar', 'valor_convertido')
```

```python
df_dolar.printSchema()
```

### Filtrando Valores

```python
# &, |, ==, ~
df.select('id', 'valor', 'chave_pix_tipo').filter(col('chave_pix_tipo') == 'cpf').show(10)
```

```python
df.select('id', 'valor', 'chave_pix_tipo').filter(col('chave_pix_tipo') != 'cpf').show(10)
```

```python
from pyspark.sql.functions import col

df.select(
    'id', 'valor', 'chave_pix_tipo'
).filter(
    (col('chave_pix_tipo') == 'cpf') &
    (col('valor').cast('double') > 10000)
).show(10)
```

```python
df.select(
    'id', 'valor', 'chave_pix_tipo'
).filter(
    (col('chave_pix_tipo') == 'cpf') |
    (col('chave_pix_tipo') == 'celular')
).show(20)
```

### Agregações e GroupBy


```python
df.select(
    'id', 'valor', 'chave_pix_tipo'
).groupBy(col('chave_pix_tipo')).count().show()
```

```python
from pyspark.sql.functions import regexp_replace, col
```

```python
df.select(
    'id', 
    col('valor').cast('double').alias('valor'), 
    'chave_pix_tipo'
).groupBy('chave_pix_tipo').sum('valor').show()
```

```python
df.withColumn('valor_cleaned', regexp_replace(col('valor'), '[^0-9.]', '').cast('double'))\
  .groupBy('chave_pix_tipo')\
  .avg('valor_cleaned')\
  .show()
```
