```python
from pyspark import SparkContext
```

```python
sc = SparkContext("local", "RDD Exemplo")
```

```python
# Leia o arquivo com a codificação UTF-8
with open("C:/Users/bianc/OneDrive/Documents/1-Estudos/2-DNC/1-Data Science/1-Material/Materia12-Modelos_de_Nuvem-e-IA/Aula 5/rdd_dataset.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

# Crie o RDD e limpe as strings em uma única operação
rdd_texto = sc.parallelize(lines).map(lambda linha: linha.strip())

# Verifique os resultados
palavras = rdd_texto.collect()
print(palavras)
```

```python
rdd_texto.collect()
```

```python
rdd_filtrado = rdd_texto.filter(lambda linha: "DNC" in linha)
```

```python
rdd_filtrado.collect()
```

```python
rdd_filtrado.count()
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, DoubleType, StringType, TimestampType
```

```python
schema_base_pix = StructType([
    StructField('id', IntegerType()),
    StructField('valor', DoubleType()),
])
```

```python
schema_base_pix = StructType([
    StructField('id', IntegerType()),
    StructField('valor', DoubleType()),
    StructField('parte_debitada_nome', StringType()),
    StructField('parte_debitada_conta', StringType()),
    StructField('parte_debitada_banco', StringType()),
    StructField('parte_creditada_nome', StringType()),
    StructField('parte_creditada_conta', StringType()),
    StructField('parte_creditada_banco', StringType()),
    StructField('chave_pix_tipo', StringType()),
    StructField('chave_pix_valor', StringType()),
    StructField('data_transacao', TimestampType())
])


df = spark.read.csv(
      path="C:/Users/bianc/OneDrive/Documents/1-Estudos/2-DNC/1-Data Science/1-Material/Materia12-Modelos_de_Nuvem-e-IA/Aula 6/base_de_dados.csv",
      header=True,
      sep=";",
      schema=schema_base_pix,
      timestampFormat="dd/MM/yyyy HH:mm"
)
```

## Manipulação de Dados I

Caso tenhamos um base con colunas variáveis

```python
import os

# Configurar Hadoop para Windows
os.makedirs("C:/hadoop/bin", exist_ok=True)
os.environ['HADOOP_HOME'] = 'C:/hadoop'
```

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Meu primeiro Dataframe").getOrCreate()
```

```python
import os
os.environ["JAVA_HOME"] = r"C:\Program Files\Eclipse Adoptium\jdk-17.0.14+7-hotspot"  # ajuste conforme instalado
os.environ["PATH"] = os.environ["JAVA_HOME"] + r"\bin;" + os.environ["PATH"]
```

```python
df = spark.read.csv(
      path="C:/Users/bianc/OneDrive/Documents/1-Estudos/2-DNC/1-Data Science/1-Material/Materia12-Modelos_de_Nuvem-e-IA/Aula 6/base_de_dados.csv",
      header=True,
      sep=";"
)
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col, to_timestamp
```

```python
df_cast = df.withColumn(
    'id', col('id').cast('int'),
).withColumn(
    'valor', col('valor').cast(DoubleType()),
).withColumn(
    'data_transacao', to_timestamp('data_transacao', "dd/MM/yyyy HH:mm"),
)
```

```python
df_cast.printSchema()
```

```python
df_cast.show()
```
