```python
def computer_col_functions(col):
    max_value = col.max()
    min_value = col.min()
    mean_value = col.mean()
    median_value = col.median()

    return pd.Series({
        "max": max_value,
        "min": min_value,
        "mean": mean_value,
        "median": median_value
    })
```

```python
df_athletes[["height", "weight", "age"]].apply(computer_col_functions)
```

```python
def create_description_column(row):
    name = row["name"]
    country = row["country_full"]

    return f"{name} | {country}"
```
