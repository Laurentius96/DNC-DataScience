**`encoding='utf-8'`:** Lida com acentos e caracteres especiais
**`decimal=','`:** Reconhece vírgula como separador decimal

## **ALTERNATIVA: Context Manager (Mais Seguro)**

```python
# 🔍 BLOCO 2 ALTERNATIVO: Com context manager
try:
    with open('ifood_df.csv', 'r', encoding='utf-8') as file:
        df = pd.read_csv(file, 
                        sep=';', 
                        decimal=',')
    print("✅ Arquivo carregado e fechado com segurança!")
except FileNotFoundError:
    print("❌ Arquivo não encontrado!")
except UnicodeDecodeError:
    print("❌ Problema de encoding - tentando latin-1...")
    df = pd.read_csv('ifood_df.csv', sep=';', encoding='latin-1')
except Exception as e:
    print(f"❌ Erro inesperado: {e}")
```

### 🧠 **Por que Context Manager é Melhor:**

**`with open()`:** Garante que arquivo seja fechado automaticamente
**Múltiplos except:** Trata diferentes tipos de erro
**Fallback encoding:** Se UTF-8 falhar, tenta latin-1

## 🎯 **PERGUNTA 1: Por que dividir por 1024?**

### **📊 Conversão de Bytes para Megabytes:**

```python
# 🔍 DEMONSTRAÇÃO: Como funciona a conversão
print("📏 CONVERSÃO DE UNIDADES DE MEMÓRIA:")
print("=" * 40)

# Exemplo com um número
bytes_exemplo = 2048576  # bytes

print(f"🔢 Valor original: {bytes_exemplo:,} bytes")
print(f"📐 1024¹ (KB): {bytes_exemplo / 1024:.2f} KB")
print(f"📐 1024² (MB): {bytes_exemplo / 1024**2:.2f} MB")
print(f"📐 1024³ (GB): {bytes_exemplo / 1024**3:.2f} GB")

print("\n🧮 EXPLICAÇÃO:")
print("1 KB = 1024 bytes")
print("1 MB = 1024 KB = 1024² bytes = 1,048,576 bytes")
print("1 GB = 1024 MB = 1024³ bytes")
```

### **🧠 Por que 1024 e não 1000?**
- **💻 Sistema binário:** Computadores trabalham em base 2
- **🔢 2¹⁰ = 1024:** Potência de 2 mais próxima de 1000
- **📏 Padrão da indústria:** Sempre usamos 1024 para memória

```python
# 🔍 BLOCO 4: Calculando PARÂMETROS POPULACIONAIS (Versão Refinada)
import pandas as pd
import numpy as np

# Primeiro, vamos carregar e verificar a base V02
df = pd.read_csv('base_ifood_limpa_Simplificada_V02.csv', sep=';')

# Verificar e converter colunas numéricas se necessário
print("🔍 VERIFICANDO TIPOS DE DADOS ORIGINAIS:")
print(df.dtypes)
print("\n" + "="*50 + "\n")

# Converter colunas que deveriam ser numéricas (se necessário)
colunas_para_converter = ['delivery_fee', 'delivery_time', 'distance', 'minimumOrderValue', 'rating']
for col in colunas_para_converter:
    if col in df.columns and df[col].dtype == 'object':
        df[col] = pd.to_numeric(df[col], errors='coerce')
        print(f"✅ Coluna '{col}' convertida para numérico")

def calcular_parametros_populacionais(df):
    """
    Calcula parâmetros populacionais com precisão profissional
    Baseado nas melhores práticas da DNC
    """
    print("\n🏢 ANÁLISE DA POPULAÇÃO COMPLETA DO iFOOD")
    print("=" * 50)
    
    # Classificação inteligente de variáveis (método DNC)
    numericas = df.select_dtypes(include=['number']).columns.tolist()
    categoricas = df.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
    
    # Remover coluna 'id' se existir nas numéricas
    if 'id' in numericas:
        numericas.remove('id')
```

```python
# 🎲 BLOCO 5: AMOSTRAGEM ALEATÓRIA E TEOREMA CENTRAL DO LIMITE
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

def criar_amostras_e_analisar_tcl(df, coluna='delivery_fee', n_amostras=1000, tamanho_amostra=30):
    """
    Cria múltiplas amostras aleatórias e demonstra o Teorema Central do Limite
    """
    print(f"🎯 DEMONSTRAÇÃO DO TEOREMA CENTRAL DO LIMITE")
    print(f"Coluna analisada: {coluna}")
    print(f"Número de amostras: {n_amostras:,}")
    print(f"Tamanho de cada amostra: {tamanho_amostra}")
    print("=" * 60)
    
    # Parâmetros populacionais (já calculados no Passo 4)
    populacao = df[coluna].dropna()
    mu_pop = populacao.mean()
    sigma_pop = populacao.std(ddof=0)
    
    print(f"📊 PARÂMETROS POPULACIONAIS:")
    print(f"   μ (média populacional): {mu_pop:.2f}")
    print(f"   σ (desvio populacional): {sigma_pop:.2f}")
    print(f"   N (tamanho população): {len(populacao):,}")
    
    # Criar múltiplas amostras e calcular suas médias
    medias_amostrais = []
    
    for i in range(n_amostras):
        amostra = populacao.sample(n=tamanho_amostra, replace=True)
        media_amostra = amostra.mean()
        medias_amostrais.append(media_amostra)
    
    medias_amostrais = np.array(medias_amostrais)
    
    # Estatísticas da distribuição amostral das médias
    media_das_medias = medias_amostrais.mean()
    erro_padrao_teorico = sigma_pop / np.sqrt(tamanho_amostra)
    erro_padrao_observado = medias_amostrais.std(ddof=1)
    
    print(f"\n🎲 DISTRIBUIÇÃO AMOSTRAL DAS MÉDIAS:")
    print(f"   x̄ (média das médias): {media_das_medias:.2f}")
    print(f"   σx̄ teórico (erro padrão): {erro_padrao_teorico:.2f}")
    print(f"   σx̄ observado: {erro_padrao_observado:.2f}")
    print(f"   Diferença: {abs(erro_padrao_teorico - erro_padrao_observado):.4f}")
    
    # Teste de normalidade
    shapiro_stat, shapiro_p = stats.shapiro(medias_amostrais[:5000])  # Shapiro max 5000
    
    print(f"\n📈 TESTE DE NORMALIDADE (Shapiro-Wilk):")
    print(f"   Estatística: {shapiro_stat:.4f}")
    print(f"   p-valor: {shapiro_p:.6f}")
    print(f"   Distribuição é normal? {'✅ SIM' if shapiro_p > 0.05 else '❌ NÃO'}")
    
    # Visualizações
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    fig.suptitle(f'Teorema Central do Limite - {coluna}', fontsize=16, fontweight='bold')
    
    # 1. Distribuição da população original
    axes[0,0].hist(populacao, bins=50, alpha=0.7, color='skyblue', edgecolor='black')
    axes[0,0].axvline(mu_pop, color='red', linestyle='--', linewidth=2, label=f'μ = {mu_pop:.2f}')
    axes[0,0].set_title('Distribuição da População Original')
    axes[0,0].set_xlabel(coluna)
    axes[0,0].set_ylabel('Frequência')
    axes[0,0].legend()
    
    # 2. Distribuição das médias amostrais
    axes[0,1].hist(medias_amostrais, bins=50, alpha=0.7, color='lightgreen', edgecolor='black')
    axes[0,1].axvline(media_das_medias, color='red', linestyle='--', linewidth=2, 
                      label=f'x̄ = {media_das_medias:.2f}')
    axes[0,1].set_title('Distribuição das Médias Amostrais')
    axes[0,1].set_xlabel(f'Médias de {coluna}')
    axes[0,1].set_ylabel('Frequência')
    axes[0,1].legend()
    
    # 3. Q-Q Plot para verificar normalidade
    stats.probplot(medias_amostrais, dist="norm", plot=axes[1,0])
    axes[1,0].set_title('Q-Q Plot (Teste Visual de Normalidade)')
    axes[1,0].grid(True)
    
    # 4. Comparação: Teórico vs Observado
    x = np.linspace(medias_amostrais.min(), medias_amostrais.max(), 100)
    distribuicao_teorica = stats.norm(mu_pop, erro_padrao_teorico)
    
    axes[1,1].hist(medias_amostrais, bins=50, density=True, alpha=0.7, 
                   color='lightgreen', label='Observado')
    axes[1,1].plot(x, distribuicao_teorica.pdf(x), 'r-', linewidth=2, 
                   label=f'Teórico N({mu_pop:.2f}, {erro_padrao_teorico:.2f})')
    axes[1,1].set_title('Comparação: Teórico vs Observado')
    axes[1,1].set_xlabel(f'Médias de {coluna}')
    axes[1,1].set_ylabel('Densidade')
    axes[1,1].legend()
    
    plt.tight_layout()
    plt.show()
    
    return {
        'medias_amostrais': medias_amostrais,
        'mu_pop': mu_pop,
        'sigma_pop': sigma_pop,
        'erro_padrao_teorico': erro_padrao_teorico,
        'erro_padrao_observado': erro_padrao_observado,
        'shapiro_p': shapiro_p
    }

# Executar demonstração do TCL
resultados_tcl = criar_amostras_e_analisar_tcl(df, coluna='delivery_fee', 
                                               n_amostras=1000, tamanho_amostra=30)
```
