## 📋 **PROJETO: "Otimização Estratégica do iFood"**

### 🏢 **CONTEXTO EMPRESARIAL**

**Situação:** Você é Cientista de Dados Jr. no iFood. A empresa quer expandir operações e otimizar custos, mas precisa tomar decisões baseadas em dados **SEM** gastar milhões em pesquisas com toda a base.

**🎯 Problema Central:** *"Como usar amostragem inteligente para tomar decisões estratégicas precisas com menor custo e risco?"*

---

## 🗺️ **ROADMAP COMPLETO - 7 ENTREGAS**

### **📦 ENTREGA 1 - Aula 1 DNC (População vs Amostra)**
**🎯 Objetivo:** Definir nossa população e entender o que queremos medir

**📋 O que vamos fazer:**
- Explorar a base completa do iFood (nossa **população**)
- Identificar **parâmetros** importantes (preço médio, tempo de entrega, rating)
- Extrair uma **amostra** e comparar **estatísticas** vs **parâmetros**

**📊 Entregável:**
```markdown
## Relatório 1: Mapeamento da População iFood
- Tamanho da população: X restaurantes
- Parâmetros-chave identificados
- Primeira amostra piloto e comparação
```

---

### **📦 ENTREGA 2 - Aula 2 DNC (Tipos de Amostragem)**
**🎯 Objetivo:** Escolher o melhor método para diferentes cenários

**📋 O que vamos fazer:**
- **Amostragem Simples:** Para análise geral de preços
- **Amostragem Estratificada:** Por categoria de comida (Pizza, Brasileira, etc.)
- **Comparar métodos:** Qual dá resultados mais representativos?

**📊 Entregável:**
```markdown
## Relatório 2: Estratégias de Amostragem
- Método recomendado para cada tipo de análise
- Código Python para cada método
- Comparação de representatividade
```

---

### **📦 ENTREGA 3 - Aula 3 DNC (Viés na Amostragem)**
**🎯 Objetivo:** Identificar e evitar armadilhas nos dados

**📋 O que vamos fazer:**
- **Detectar viés geográfico:** Restaurantes só de uma região?
- **Identificar viés de preço:** Só restaurantes caros/baratos?
- **Corrigir vieses:** Técnicas para amostra mais justa

**📊 Entregável:**
```markdown
## Relatório 3: Auditoria de Vieses
- Lista de vieses identificados na base
- Impacto de cada viés nas decisões
- Plano de correção implementado
```

---

### **📦 ENTREGA 4 - Aula 4 DNC (Distribuições Amostrais)**
**🎯 Objetivo:** Entender a variabilidade das nossas estimativas

**📋 O que vamos fazer:**
- Simular **100 amostras** diferentes
- Plotar distribuição das **médias de preço**
- Calcular **intervalo de confiança** para nossas estimativas

**📊 Entregável:**
```markdown
## Relatório 4: Análise de Variabilidade
- Gráfico da distribuição das médias amostrais
- Intervalo de confiança para parâmetros-chave
- Recomendação de precisão esperada
```

---

### **📦 ENTREGA 5 - Aula 5 DNC (Teorema do Limite Central)**
**🎯 Objetivo:** Validar que podemos usar distribuição normal

**📋 O que vamos fazer:**
- Testar TLC com dados de **tempo de entrega** (não-normal)
- Mostrar como médias ficam normais mesmo com dados "estranhos"
- Aplicar testes estatísticos baseados em normalidade

**📊 Entregável:**
```markdown
## Relatório 5: Validação Estatística
- Prova visual do TLC com nossos dados
- Justificativa para usar testes paramétricos
- Limites de aplicação do TLC
```

---

### **📦 ENTREGA 6 - Aula 6 DNC (Tamanho Amostral)**
**🎯 Objetivo:** Calcular o tamanho ideal para cada análise

**📋 O que vamos fazer:**
- **Cenário A:** Pesquisa de satisfação (quantos clientes entrevistar?)
- **Cenário B:** Teste A/B de nova interface (quantos usuários?)
- **Análise custo-benefício:** Precisão vs investimento

**📊 Entregável:**
```markdown
## Relatório 6: Dimensionamento Estratégico
- Tamanho amostral para cada tipo de pesquisa
- Calculadora Python para futuros projetos
- Análise de ROI por tamanho de amostra
```

---

### **📦 ENTREGA 7 - Aula 7 DNC (Avaliação em Python)**
**🎯 Objetivo:** Consolidar tudo em um experimento real

**📋 O que vamos fazer:**
- **Experimento A/B:** Testar nova estratégia de recomendação
- **Aplicar tudo:** Amostragem + controle de viés + tamanho ideal
- **Dashboard final:** Resultados e recomendações

**📊 Entregável:**
```markdown
## Relatório Final: Experimento Completo
- Metodologia completa aplicada
- Resultados do teste A/B
- Recomendações estratégicas para o iFood
- Toolkit Python para futuros projetos
```
