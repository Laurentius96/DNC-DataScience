```python
# 🔍 BLOCO 2: Carregamento PROFISSIONAL dos dados
import os

# Verificar se arquivo existe
if os.path.exists('ifood_df.csv'):
    try:
        # Carregamento seguro com parâmetros brasileiros
        df = pd.read_csv('ifood_df.csv', 
                        sep=';',           # Separador brasileiro
                        encoding='utf-8',  # Acentos brasileiros
```

### 🧠 **Anatomia do Código Profissional:**

**`try/except`:** Captura erros de forma elegante
**`sep=';'`:** Define separador como ponto-e-vírgula (padrão brasileiro)
**`encoding='utf-8'`:** Lida com acentos e caracteres especiais
**`decimal=','`:** Reconhece vírgula como separador decimal

## **ALTERNATIVA: Context Manager (Mais Seguro)**

```python
# 🔍 BLOCO 2 ALTERNATIVO: Com context manager
try:
    with open('ifood_df.csv', 'r', encoding='utf-8') as file:
        df = pd.read_csv(file, 
                        sep=';', 
                        decimal=',')
    print("✅ Arquivo carregado e fechado com segurança!")
except FileNotFoundError:
    print("❌ Arquivo não encontrado!")
except UnicodeDecodeError:
    print("❌ Problema de encoding - tentando latin-1...")
    df = pd.read_csv('ifood_df.csv', sep=';', encoding='latin-1')
except Exception as e:
    print(f"❌ Erro inesperado: {e}")
```

### 🧠 **Por que Context Manager é Melhor:**

**`with open()`:** Garante que arquivo seja fechado automaticamente
**Múltiplos except:** Trata diferentes tipos de erro
**Fallback encoding:** Se UTF-8 falhar, tenta latin-1

## 🎯 **PERGUNTA 1: Por que dividir por 1024?**

### **📊 Conversão de Bytes para Megabytes:**

```python
# 🔍 DEMONSTRAÇÃO: Como funciona a conversão
print("📏 CONVERSÃO DE UNIDADES DE MEMÓRIA:")
print("=" * 40)

# Exemplo com um número
bytes_exemplo = 2048576  # bytes

print(f"🔢 Valor original: {bytes_exemplo:,} bytes")
print(f"📐 1024¹ (KB): {bytes_exemplo / 1024:.2f} KB")
print(f"📐 1024² (MB): {bytes_exemplo / 1024**2:.2f} MB")
print(f"📐 1024³ (GB): {bytes_exemplo / 1024**3:.2f} GB")

print("\n🧮 EXPLICAÇÃO:")
print("1 KB = 1024 bytes")
print("1 MB = 1024 KB = 1024² bytes = 1,048,576 bytes")
print("1 GB = 1024 MB = 1024³ bytes")
```

### **🧠 Por que 1024 e não 1000?**
- **💻 Sistema binário:** Computadores trabalham em base 2
- **🔢 2¹⁰ = 1024:** Potência de 2 mais próxima de 1000
- **📏 Padrão da indústria:** Sempre usamos 1024 para memória

## 🎯 **PERGUNTA 2: Por que dois .sum()?**

### **📊 Demonstração do Duplo .sum():**

```python
# 🔍 DEMONSTRAÇÃO: Por que precisamos de dois .sum()
import pandas as pd
```

### **🧠 Anatomia do Duplo .sum():**

- **`df.isnull()`:** Retorna matriz True/False para cada célula
- **`df.isnull().sum()`:** Primeiro .sum() conta nulos **por coluna**
- **`df.isnull().sum().sum()`:** Segundo .sum() soma **todas as colunas**

```python
import random

def criar_amostra_aleatoria(df, tamanho_amostra=100, seed=42):
    """
    Cria amostra aleatória simples e calcula estatísticas amostrais
    Compara com parâmetros populacionais
    """
    # Definir seed para reprodutibilidade
    random.seed(seed)
    np.random.seed(seed)
    
    print("🎲 CRIANDO AMOSTRA ALEATÓRIA SIMPLES")
    print("=" * 50)
    
    # Criar amostra aleatória
    amostra = df.sample(n=tamanho_amostra, random_state=seed)
    
    print(f"📊 Tamanho da população: {len(df):,}")
    print(f"🎯 Tamanho da amostra: {len(amostra):,}")
    print(f"📈 Percentual amostrado: {(len(amostra)/len(df)*100):.2f}%")
    print(f"🔢 Seed utilizada: {seed} (para reprodutibilidade)")
    
    # Identificar variáveis numéricas
    numericas = df.select_dtypes(include=['number']).columns.tolist()
    
    print(f"\n📋 ESTATÍSTICAS AMOSTRAIS vs POPULACIONAIS:")
    print("=" * 60)
    
    # Comparar estatísticas
    comparacao = {}
    for col in numericas:
        # População (parâmetros verdadeiros)
        pop_dados = df[col].dropna()
        
        # Amostra (estimativas)
        amostra_dados = amostra[col].dropna()
        
        comparacao[col] = {
            'População': {
                'N': len(pop_dados),
                'μ (média)': pop_dados.mean(),
                'σ (desvio padrão)': pop_dados.std(ddof=0),  # População: ddof=0
                'Mediana': pop_dados.median()
            },
            'Amostra': {
                'n': len(amostra_dados),
                'x̄ (média amostral)': amostra_dados.mean(),
                's (desvio amostral)': amostra_dados.std(ddof=1),  # Amostra: ddof=1
                'Mediana': amostra_dados.median()
            }
        }
        
        # Calcular erros de estimação
        erro_media = abs(comparacao[col]['Amostra']['x̄ (média amostral)'] - 
                        comparacao[col]['População']['μ (média)'])
        erro_percentual = (erro_media / comparacao[col]['População']['μ (média)']) * 100
        
        comparacao[col]['Erro'] = {
            'Erro absoluto': erro_media,
            'Erro percentual': erro_percentual
        }
    
    # Exibir comparação elegante
    for coluna, stats in comparacao.items():
        print(f"\n📈 {coluna.upper()}:")
        print("   POPULAÇÃO (Parâmetros Verdadeiros):")
        for nome, valor in stats['População'].items():
```
