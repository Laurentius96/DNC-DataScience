# 🌟 **PASSO 2 CORRIGIDO: Carregamento Profissional**

## 🎯 **Problemas Identificados e Soluções:**

1. **🔍 Separador:** CSV brasileiro usa ";" ao invés de ","
2. **🛡️ Segurança:** Devemos abrir/fechar arquivos com segurança
3. **📊 Encoding:** Arquivos brasileiros podem ter acentos (UTF-8)

---

## **VERSÃO PROFISSIONAL:**

```python
# 🔍 BLOCO 2: Carregamento PROFISSIONAL dos dados
import os

# Verificar se arquivo existe
if os.path.exists('ifood_df.csv'):
    try:
        # Carregamento seguro com parâmetros brasileiros
        df = pd.read_csv('ifood_df.csv', 
                        sep=';',           # Separador brasileiro
                        encoding='utf-8',  # Acentos brasileiros
                        decimal=',')       # Decimal brasileiro
        print("✅ Arquivo carregado com sucesso!")
    except Exception as e:
        print(f"❌ Erro ao carregar arquivo: {e}")
else:
    print("❌ Arquivo 'ifood_df.csv' não encontrado!")
```

### 🧠 **Anatomia do Código Profissional:**

**`os.path.exists()`:** Verifica se arquivo existe antes de tentar abrir
**`try/except`:** Captura erros de forma elegante
**`sep=';'`:** Define separador como ponto-e-vírgula (padrão brasileiro)
**`encoding='utf-8'`:** Lida com acentos e caracteres especiais
**`decimal=','`:** Reconhece vírgula como separador decimal

---

## **ALTERNATIVA: Context Manager (Mais Seguro)**

```python
# 🔍 BLOCO 2 ALTERNATIVO: Com context manager
try:
    with open('ifood_df.csv', 'r', encoding='utf-8') as file:
        df = pd.read_csv(file, 
                        sep=';', 
                        decimal=',')
    print("✅ Arquivo carregado e fechado com segurança!")
except FileNotFoundError:
    print("❌ Arquivo não encontrado!")
except UnicodeDecodeError:
    print("❌ Problema de encoding - tentando latin-1...")
    df = pd.read_csv('ifood_df.csv', sep=';', encoding='latin-1')
except Exception as e:
    print(f"❌ Erro inesperado: {e}")
```

### 🧠 **Por que Context Manager é Melhor:**

**`with open()`:** Garante que arquivo seja fechado automaticamente
**Múltiplos except:** Trata diferentes tipos de erro
**Fallback encoding:** Se UTF-8 falhar, tenta latin-1

---

# 🌟 **PASSO 4: Primeira Exploração da População (Versão Profissional)**

Agora vamos explorar nossa população com **análise crítica profissional**:

```python
# 🔍 BLOCO 3: Exploração PROFISSIONAL da população
def explorar_dataset(df):
    """
    Função para exploração completa e profissional do dataset
    """
    print("🏢 POPULAÇÃO COMPLETA DO iFOOD")
    print("=" * 50)
    
    # Informações básicas
    print(f"📊 Total de restaurantes (população): {len(df):,}")
    print(f"📋 Total de características (colunas): {len(df.columns)}")
    print(f"💾 Uso de memória: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    
    # Verificação de qualidade dos dados
    print(f"\n🔍 QUALIDADE DOS DADOS:")
    print(f"❌ Valores nulos: {df.isnull().sum().sum()}")
    print(f"🔄 Linhas duplicadas: {df.duplicated().sum()}")
    
    # Tipos de dados
    print(f"\n📝 TIPOS DE DADOS:")
    for col, dtype in df.dtypes.items():
        null_count = df[col].isnull().sum()
        print(f"   {col}: {dtype} (nulos: {null_count})")
    
    return df.head()

# Executar exploração
primeiras_linhas = explorar_dataset(df)
print(f"\n🔍 AMOSTRA DOS DADOS:")
print(primeiras_linhas)
```

### 🧠 **Anatomia do Código Profissional:**

**`def explorar_dataset(df):`** Função reutilizável (DRY principle)
**`{len(df):,}`** Formatação com separador de milhares
**`memory_usage(deep=True)`** Análise real do uso de memória
**`df.isnull().sum().sum()`** Contagem total de valores nulos
**`df.duplicated().sum()`** Detecção de dados duplicados
**`df.dtypes.items()`** Análise dos tipos de dados por coluna
