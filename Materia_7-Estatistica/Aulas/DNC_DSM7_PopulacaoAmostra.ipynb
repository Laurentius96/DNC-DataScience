## **ALTERNATIVA: Context Manager (Mais Seguro)**

```python
# 🔍 BLOCO 2 ALTERNATIVO: Com context manager
try:
    with open('ifood_df.csv', 'r', encoding='utf-8') as file:
        df = pd.read_csv(file, 
                        sep=';', 
                        decimal=',')
    print("✅ Arquivo carregado e fechado com segurança!")
except FileNotFoundError:
    print("❌ Arquivo não encontrado!")
except UnicodeDecodeError:
    print("❌ Problema de encoding - tentando latin-1...")
    df = pd.read_csv('ifood_df.csv', sep=';', encoding='latin-1')
except Exception as e:
    print(f"❌ Erro inesperado: {e}")
```

### 🧠 **Por que Context Manager é Melhor:**

**`with open()`:** Garante que arquivo seja fechado automaticamente
**Múltiplos except:** Trata diferentes tipos de erro
**Fallback encoding:** Se UTF-8 falhar, tenta latin-1

## 🎯 **PERGUNTA 1: Por que dividir por 1024?**

### **📊 Conversão de Bytes para Megabytes:**

### **🧠 Por que 1024 e não 1000?**
- **💻 Sistema binário:** Computadores trabalham em base 2
- **🔢 2¹⁰ = 1024:** Potência de 2 mais próxima de 1000
- **📏 Padrão da indústria:** Sempre usamos 1024 para memória

# 🌟 **PASSO 5: Identificando Parâmetros Populacionais**

Agora vamos calcular os **parâmetros verdadeiros** da nossa população (não estimativas!):

```python
# 🔍 BLOCO 4: Calculando PARÂMETROS POPULACIONAIS
def calcular_parametros_populacionais(df, coluna_numerica):
    """
    Calcula os parâmetros verdadeiros da população
    (não são estimativas - são valores REAIS!)
    """
    print(f"🎯 PARÂMETROS POPULACIONAIS - {coluna_numerica.upper()}")
    print("=" * 50)
    
    # Limpeza profissional dos dados
    dados_limpos = df[coluna_numerica].dropna()
    valores_removidos = len(df) - len(dados_limpos)
    
    if valores_removidos > 0:
        print(f"🧹 Valores nulos removidos: {valores_removidos}")
        print(f"📊 Dados válidos: {len(dados_limpos):,}")
    
    # PARÂMETROS POPULACIONAIS (símbolos gregos!)
    parametros = {
        'μ (mu)': dados_limpos.mean(),           # Média populacional
        'σ² (sigma²)': dados_limpos.var(ddof=0), # Variância populacional  
        'σ (sigma)': dados_limpos.std(ddof=0),   # Desvio padrão populacional
        'Mediana': dados_limpos.median(),         # Mediana populacional
        'Min': dados_limpos.min(),               # Mínimo
        'Max': dados_limpos.max(),               # Máximo
        'N': len(dados_limpos)                   # Tamanho populacional
    }
    
    # Exibição profissional
    for nome, valor in parametros.items():
        if isinstance(valor, (int, float)) and nome != 'N':
            print(f"📈 {nome:<12}: {valor:,.2f}")
        else:
            print(f"📈 {nome:<12}: {valor:,}")
    
    return parametros
```

### 🧠 **Anatomia do Código Profissional:**

- **`dropna()`:** Remove valores nulos de forma segura
- **`ddof=0`:** Divisor populacional (N) ao invés de amostral (N-1)
- **`select_dtypes(include=[np.number])`:** Identifica automaticamente colunas numéricas
- **Símbolos gregos:** μ (mu), σ (sigma) - notação estatística correta
- **Tratamento de erros:** try/except para robustez

````python
# 🔍 BLOCO 5: Calculando PARÂMETROS POPULACIONAIS
def calcular_parametros_populacionais(df, coluna_numerica):
    """
    Calcula os parâmetros verdadeiros da população
    (não são estimativas - são valores REAIS!)
    """
    print(f"🎯 PARÂMETROS POPULACIONAIS - {coluna_numerica.upper()}")
    print("=" * 50)
    
    # Limpeza profissional dos dados
    dados_limpos = df[coluna_numerica].dropna()
    valores_removidos = len(df) - len(dados_limpos)
    
    if valores_removidos > 0:
        print(f"🧹 Valores nulos removidos: {valores_removidos}")
        print(f"📊 Dados válidos: {len(dados_limpos):,}")
    
    # PARÂMETROS POPULACIONAIS (símbolos gregos!)
    parametros = {
        'μ (mu)': dados_limpos.mean(),           # Média populacional
        'σ² (sigma²)': dados_limpos.var(ddof=0), # Variância populacional  
        'σ (sigma)': dados_limpos.std(ddof=0),   # Desvio padrão populacional
        'Mediana': dados_limpos.median(),         # Mediana populacional
        'Min': dados_limpos.min(),               # Mínimo
        'Max': dados_limpos.max(),               # Máximo
        'N': len(dados_limpos)                   # Tamanho populacional
    }
    
    # Exibição profissional
    for nome, valor in parametros.items():
        if isinstance(valor, (int, float)) and nome != 'N':
            print(f"📈 {nome:<12}: {valor:,.2f}")
        else:
            print(f"📈 {nome:<12}: {valor:,}")
    
    return parametros

# Exemplo com uma coluna numérica (assumindo que existe 'preco' ou similar)
# ATENÇÃO: Substitua 'preco' pela coluna numérica real do seu dataset
try:
    # Verificar quais colunas numéricas existem
    colunas_numericas = df.select_dtypes(include=[np.number]).columns.tolist()
    print(f"🔢 Colunas numéricas disponíveis: {colunas_numericas}")
    
    if colunas_numericas:
        # Usar a primeira coluna numérica encontrada
        coluna_escolhida = colunas_numericas[0]
        parametros = calcular_parametros_populacionais(df, coluna_escolhida)
    else:
        print("❌ Nenhuma coluna numérica encontrada!")
        
except Exception as e:
    print(f"❌ Erro: {e}")
    print("💡 Verifique se o dataset foi carregado corretamente")
````
