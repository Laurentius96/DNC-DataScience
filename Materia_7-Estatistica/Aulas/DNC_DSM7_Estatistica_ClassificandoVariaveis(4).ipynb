### 1. 🧠 Ficha de Revisão Rápida | (O Essencial)

<br>

> - **Variável Numérica:** valores contínuos ou discretos (`float`, `int`).  
> - **Variável Categórica:** rótulos fixos (`object` ou `category`).  
> - **Variável Ordinal:** categorias com ordem natural (ex: `price_range`).  
> - **Variável Texto:** strings livres (ex: `name`, `url`).  
> - **Variável Geoespacial:** pares de latitude/longitude.  
> - Use `df.select_dtypes()` para selecionar por tipo.

<br>


### 2. 🔬 Mergulho Profundo | (Os Detalhes)

<br>

#### **🎯 O Conceito Central**  
Classificar variáveis é **identificar** cada coluna pelo tipo de dado que armazena, para **aplicar** as transformações corretas:

1. **Numéricas:** permitem estatísticas e modelagem direta.  
2. **Categóricas:** exigem codificação (*one-hot*, *label encoding*).  
3. **Ordinais:** podem usar mapeamento numérico preservando ordem.  
4. **Texto/URL:** requerem técnicas de NLP ou extração de *features*.  
5. **Geo:** podem gerar distâncias ou clusters.

<br>

#### **🔗 Analogia de Data Science**  
Imagine um buffet onde cada prato (coluna) pede um talher apropriado:  
- Colher para sopas (*numérico*),  
- Garfo para saladas (*categórico*),  
- Faca para carnes (*ordinal*),  
- Pinça para petiscos (*texto*),  
- Fita métrica para medir lotes (*geo*).  

Se usar o talher errado, você não aproveita o prato.

#### **Nível Intermediário: Conversão e Codificação de Categóricas/Ordinais**


````
import pandas as pd

df = pd.read_csv("/content/ifood-restaurants-november-2020.csv")

# Ordinal: price_range
order = ["CHEAPEST","CHEAP","MODERATE","EXPENSIVE","MOST_EXPENSIVE"]
df["price_ord"] = pd.Categorical(df["price_range"], categories=order, ordered=True)

# Label encoding
df["price_code"] = df["price_ord"].cat.codes + 1
print(df["price_code"].value_counts().sort_index())
````

#### **Nível DEUS (1/3): Detecção Avançada de Tipos e Resumo Programático**


````
import pandas as pd, numpy as np

df = pd.read_csv("/content/ifood-restaurants-november-2020.csv")

summary = []
for col in df.columns:
    dtype = df[col].dtype
    unique = df[col].nunique(dropna=False)
    kind = "num" if np.issubdtype(dtype, np.number) else "cat" if unique<20 else "text"
    summary.append((col, str(dtype), unique, kind))

summary_df = pd.DataFrame(summary, columns=["coluna","dtype","n_unique","tipo_estimado"])
print(summary_df)
````

````
# Pratique seu código aqui!
````

* **O que o código faz:**  

  **1) Explicação Linha a Linha:**  
  - Para cada coluna, captura `dtype` e conta valores únicos.  
  - Usa regra simples para estimar “num”, “cat” ou “text”.  
  - Monta `DataFrame` resumo.  

  **2) Tabela de Estados Intermediários:**

  ```markdown
  | Passo | Expressão               | Saída         |
  |:-----:|:------------------------|:--------------|
  | 1     | `dtype`                 | numpy dtype   |
  | 2     | `nunique(dropna=False)` | Inteiro       |
  | 3     | `kind=...`              | String tipo   |
  ```

  **3) Diagrama Mental:**  
  Um robô inspeciona colunas e rotula como planta, arbusto ou árvore.

* **Cenário de Mercado:**  
  Automatiza análise inicial de datasets em pipelines.

* **Boas Práticas:**  
  Ajuste thresholds de `unique` conforme domínio dos dados.


#### **Nível DEUS (2/3): Pré-processamento Automático com ColumnTransformer**


````
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import numpy as np

df = pd.read_csv("/content/ifood-restaurants-november-2020.csv")
num_feats = df.select_dtypes(include=[np.number]).columns.tolist()
cat_feats = df.select_dtypes(include=["object"]).columns.tolist()

pre = ColumnTransformer([
  ("num", StandardScaler(), num_feats),
  ("cat", OneHotEncoder(handle_unknown="ignore"), cat_feats)
])
X = pre.fit_transform(df)
print("Shape pós-transform:", X.shape)
````

````
# Pratique seu código aqui!
````

* **O que o código faz:**  

  - Separa colunas numéricas e categóricas.  
  - Aplica escala em números e one-hot em categorias.  
  - Retorna matriz pronta para modelo.  

* **Tabela de Estados Intermediários:**

  ```markdown
  | Passo | Expressão             | Saída           |
  |:-----:|:----------------------|:----------------|
  | 1     | `ColumnTransformer`   | Objeto préproc  |
  | 2     | `fit_transform(df)`   | array numpy     |
  ```

* **Diagrama Mental:**  
  Linha de montagem aplica processos e junta tudo no final.

* **Cenário de Mercado:**  
  Padrão em pipelines de ML para produção.

* **Boas Práticas:**  
  Mantenha transformações em `ColumnTransformer` para evitar *data leakage*.


#### **Nível DEUS (3/3): Pipeline Completo para Modelagem**


````
import pandas as pd, numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("/content/ifood-restaurants-november-2020.csv")
X = df.drop("price_range", axis=1)
y = LabelEncoder().fit_transform(df["price_range"])

num_feats = X.select_dtypes(include=[np.number]).columns
cat_feats = X.select_dtypes(include=["object"]).columns

pre = ColumnTransformer([
  ("num", StandardScaler(), num_feats),
  ("cat", OneHotEncoder(handle_unknown="ignore"), cat_feats)
])

pipeline = Pipeline([
  ("preprocessor", pre),
  ("clf", RandomForestClassifier(n_estimators=50, random_state=0))
])

pipeline.fit(X, y)
print("Acurácia:", pipeline.score(X, y))
````

````
# Pratique seu código aqui!
````

* **O que o código faz:**  

  1. Lê dados e separa `X` e `y`.  
  2. Define transformações numéricas e categóricas.  
  3. Monta pipeline com pré-processamento e classificador.  
  4. Treina e exibe acurácia.  

* **Tabela de Estados Intermediários:**

  ```markdown
  | Passo         | Expressão       | Saída      |
  |:-------------:|:----------------|:-----------|
  | 1             | `Pipeline([...])`| Pipeline   |
  | 2             | `fit(X,y)`      | Modelo ajustado |
  | 3             | `score(X,y)`    | Float      |
  ```

* **Diagrama Mental:**  
Esteira industrial que limpa, transforma e monta o produto (modelo).

* **Cenário de Mercado:**  
Recomendação de restaurantes baseada em atributos operacionais.

* **Boas Práticas:**  
Separe pré-processamento e modelo para deploy seguro.
