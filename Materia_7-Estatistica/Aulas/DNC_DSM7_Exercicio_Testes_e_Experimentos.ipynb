## 📊 Cálculo de Taxa de Conversão e Impacto de Mudanças

> Uma empresa de tecnologia quer avaliar o impacto de uma nova interface em seu aplicativo. Foi realizado um experimento onde um grupo de usuários utilizou a versão antiga (grupo controle) e outro grupo usou a nova versão (grupo de teste).
>
> Os resultados coletados foram os seguintes:

<br>

<center>

| Grupo | Número de Usuários | Número de Ações Completadas | Taxa de Conversão (%) |
| :---: | :---: | :---: | :---: |
| **Controle** | 5.000 | 1.200 | **?** |
| **Teste** | 5.000 | 1.500 | **?** |

</center>

<br>

-----

### 🎯 Desafios:

  * **a)** Calcule a taxa de conversão para cada grupo.
  * **b)** Determine o percentual de aumento ou redução na conversão com a nova interface.
  * **c)** Se o critério para adoção da nova interface for um aumento mínimo de **20% na taxa de conversão**, a mudança deve ser implementada?

```python
numero_usuario_control = 5000
numero_usuario_teste = 5000
numero_acao_comple_cont = 1200
numero_acao_comple_test = 1500

import pandas as pd
```

**a)** Calcule a taxa de conversão para cada grupo.

```python
taxa_conv_controle = (numero_acao_comple_cont/numero_usuario_control)
taxa_conv_teste = (numero_acao_comple_test/numero_usuario_teste)

print(f"Taxa de conversão Controle: {taxa_conv_controle:.0%}")
print(f"Taxa de conversão Teste: {taxa_conv_teste:.0%}")
```

**b)** Determine o percentual de aumento ou redução na conversão com a nova interface.

```python
percentual_mudanca = (taxa_conv_teste-taxa_conv_controle)/taxa_conv_controle
print(f"Taxa de converção: {percentual_mudanca:.0%}")
```

**c)** Se o critério para adoção da nova interface for um aumento mínimo de **20% na taxa de conversão**, a mudança deve ser implementada?

```python
if percentual_mudanca >= 0.20:
    print("Implementar mudança de interface")
else:
    print("Não implementar mudança de interface")
```

## 📈 Exercício 2: Teste A/B e Análise de Resultados

> Uma empresa de comércio eletrônico está realizando um teste A/B para avaliar qual versão de um anúncio gera mais cliques. A empresa coletou os seguintes dados:

<br>

<center>

| Versão do Anúncio | Impressões | Cliques | Taxa de Cliques (%) |
| :---: | :---: | :---: | :---: |
| **Anúncio A** | 50.000 | 2.000 | **?** |
| **Anúncio B** | 50.000 | 2.300 | **?** |

</center>

<br>

### 🎯 Desafios:

* **a)** Calcule a taxa de cliques para cada versão do anúncio.
* **b)** Qual versão teve melhor desempenho?
* **c)** Se o nível de significância adotado para o experimento for **5% (α=0.05)** e o **p-valor** calculado for **0,08**, podemos afirmar que a diferença entre os anúncios é estatisticamente significativa?
---

```python
impressoes_A = 50000
impressoes_B = 50000
cliques_A = 2000
cliques_B = 2300
```

**a)** Calcule a taxa de cliques para cada versão do anúncio.

```python
taxa_de_cliques_A = cliques_A/impressoes_A 
taxa_de_cliques_B = cliques_B/impressoes_B

print(f"Taxa de cliques A: {taxa_de_cliques_A:.1%}")
print(f"Taxa de cliques B: {taxa_de_cliques_B:.1%}")
```

**b)** Qual versão teve melhor desempenho?

```python
print(f"b) Avaliação de Desempenho:")
if taxa_de_cliques_B > taxa_de_cliques_A:
    print(f"   - O Anúncio B teve um desempenho melhor ({taxa_de_cliques_B:.2%}) do que o Anúncio A ({taxa_de_cliques_A:.2%}).")
elif taxa_de_cliques_A > taxa_de_cliques_B:
    print(f"   - O Anúncio A teve um desempenho melhor ({taxa_de_cliques_A:.2%}) do que o Anúncio B ({taxa_de_cliques_B:.2%}).")
else:
    print("   - Ambos os anúncios tiveram o mesmo desempenho.")
```

**c)** Se o nível de significância adotado para o experimento for **5% (α=0.05)** e o **p-valor** calculado for **0,08**, podemos afirmar que a diferença entre os anúncios é estatisticamente significativa?

```python
p_valor_do_teste = 0.08
nivel_de_significancia = 0.05

print(f"   - P-Valor do teste: {p_valor_do_teste}")
print(f"   - Nível de Significância (α): {nivel_de_significancia}")

# A "Regra de Ouro" da estatística:
# Se o p-valor for MENOR ou IGUAL ao nível de significância, o resultado é significativo.
if p_valor_do_teste <= nivel_de_significancia:
    print("\n   -> Conclusão: SIM, a diferença é estatisticamente significativa.")
    print("      (Podemos afirmar com confiança que o Anúncio B é realmente melhor.)")
else:
    print("\n   -> Conclusão: NÃO, a diferença NÃO é estatisticamente significativa.")
    print("      (Apesar do Anúncio B parecer melhor, não temos evidências fortes o suficiente.")
    print("      (A diferença pode ter ocorrido apenas por sorte ou acaso.)")
```

## 🤖 Exercício 3: Interpretação de Teste Estatístico

> Uma empresa realizou um teste para verificar se a inclusão de um chatbot em seu site aumentaria a taxa de fechamento de vendas. Os resultados foram analisados utilizando um teste t de Student, e os seguintes valores foram obtidos:
>
> * **Média de vendas no grupo controle:** 5,2
> * **Média de vendas no grupo teste:** 6,1
> * **Desvio padrão:** 1,4
> * **Número de participantes em cada grupo:** 1.000
> * **p-valor:** 0,03

<br>

### 🎯 Desafios:

* **a)** O aumento na média de vendas foi estatisticamente significativo considerando um nível de significância de **5%**?
* **b)** Explique o significado do p-valor e como ele deve ser interpretado.
---

```python
# p-valor: O resultado que o teste estatístico nos deu.
p_valor = 0.03

# nível_significancia (Alfa): O nosso "limite de dúvida" aceitável.
# O problema definiu como 5% (ou 0.05).
nivel_significancia = 0.05

print(f"--- Análise do Teste do Chatbot ---")
print(f"P-Valor obtido: {p_valor}")
print(f"Nível de Significância (α) definido: {nivel_significancia}")
print("-" * 100 + "\n")

# Esta é a "Regra de Ouro" da estatística:
# Se o P-Valor for MENOR ou IGUAL ao Nível de Significância,
# nós rejeitamos a "sorte" e aceitamos que o resultado é REAL.

if p_valor <= nivel_significancia:
    
    print("a) Conclusão: SIM, o aumento na média de vendas é estatisticamente significativo.")
    print("   (A evidência sugere que o Chatbot realmente causou o aumento nas vendas.)")
else:
   
    print("a) Conclusão: NÃO, o aumento na média de vendas NÃO é estatisticamente significativo.")
    print("   (Não podemos afirmar que o Chatbot causou o aumento; a diferença pode ter sido apenas sorte.)")

```

**b)** Explique o significado do p-valor e como ele deve ser interpretado.

```python
print(f"""
   O p-valor ({p_valor}, ou 3%) representa a probabilidade de obter um
   resultado tão extremo quanto o observado (um aumento de 5.2 para 6.1),
   partindo do princípio de que o chatbot NÃO tivesse efeito nenhum
   (ou seja, se a 'hipótese nula' fosse verdadeira).
""")
print("   Em termos simples: É a 'probabilidade de ter sido apenas sorte'.")
print("   Como a chance de ser sorte (3%) é muito baixa (menor que nosso limite de 5%),")
print("   nós concluímos que NÃO foi sorte.")
```
